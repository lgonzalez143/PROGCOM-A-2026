````python
class disfraz():
  #creando el constructor
  def __init__(self, color, talla, forma, material, personaje):
    #creando atributos de clase
    self.color=color
    self.talla=talla
    self.forma=forma
    self.material=material
    self.personaje=personaje

  def imprimir(self):
    return f"El disfraz es de {self.personaje}. \nEs de talla {self.talla}.\nEs de color {self.color}.\nEs de forma {self.forma}.\nEs de material {self.material}."
````

````python
perro=disfraz("amarillo con marrón","10","cuadrado","plástico","Bob Esponja")
````

````python
print(perro.imprimir())
````

````python
class Edificio():
  #construyendo el constructor
  def __init__(self, material, ubicación, fachada, ventana, puertas, nombre):
    self.material=material
    self.ubicación=ubicación
    self.fachada=fachada
    self.ventana=ventana
    self.puertas=puertas
    self.nombre=nombre

  def direccion(tipo):
    return f"Tipo de edificación:\nDirección:{self.ubicación}"

  def descripción(self,tipo):
    print(f"{self.nombre} es un {tipo} ubicado en {self.ubicación}")
    print(f"Tiene {self.ventana} ventanas, {self.puertas} puertas")
    print(f"Su fachada es {self.fachada}")
````

````python
colegio=Edificio("Ladrillo","Bucaramanga","De colores pasteles","2","1","los angelitos")
floristeria=Edificio("Cemento","Floridablanca","Color negro","3","2","Florecitas")
````
