```python
#Ejercicio 5
notas = {"Ana":4.5,
         "Andrés":3.8,
         "Manuela":4.2}
def promedio_total(dict_notas):
  return sum(dict_notas.values())/len(dict_notas)

def nota_estudiantes(dict_notas, nombre):
  return dict_notas.get(nombre, "Estudiante no encontrado")

  print(f"Promedio: {promedio_total(notas)}")
  print(f"Nota de Ana: {nota_estudiantes(notas, 'Ana')}")
```

```python
#Ejercicio 7
palabra = "magia"
conteo ={}
for letra in palabra:
  conteo[letra] = conteo.get(letra, 0) + 1
print(conteo)
```

```python
#Ejercicio 8
items = ["gema", "oro", "gema",
         "madera", "oro", "madera"]
conteo = {}
for item in items:
  conteo[item] = conteo.get(item, 0) + 1
print(conteo)
```
