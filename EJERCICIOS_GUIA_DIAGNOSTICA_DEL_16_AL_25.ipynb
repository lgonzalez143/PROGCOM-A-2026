```python
n_str = input("Por favor, introduce un número entero positivo para calcular la suma hasta él: ")

if n_str.isdigit():
    n = int(n_str)
    if n > 0:
        suma = 0
        for i in range(1, n + 1):
            suma += i
        print(f"La suma de los números naturales hasta {n} es: {suma}")
    else:
        print("Entrada no válida. introduce un número entero positivo.")
else:
    print("Entrada no válida. introduce un número entero positivo.")
```

```python
numero_str = input("Por favor, introduce un número entero no negativo para calcular su factorial: ")

try:
    numero = int(numero_str)
    if numero < 0:
        print("Entrada no válida, introduce un número entero no negativo.")
    elif numero == 0:
        print(f"El factorial de {numero} es: 1")
    else:
        factorial = 1
        for i in range(1, numero + 1):
            factorial *= i
        print(f"El factorial de {numero} es: {factorial}")
except ValueError:
    print("Entrada no válida, introduce un número entero no negativo.")
```

```python
n_fib_str = input("Por favor, introduce un número entero positivo para la serie Fibonacci: ")
if n_fib_str.isdigit():
    n_fib = int(n_fib_str)
    if n_fib <= 0:
        print("Entrada no válida. introduce un número entero positivo.")
    elif n_fib == 1:
        print("Serie Fibonacci hasta el 1er número:")
        print("0")
    else:
        print(f"Serie Fibonacci hasta los primeros {n_fib} números:")
        a, b = 0, 1
        print(a)
        for _ in range(n_fib - 1):
            print(b)
            a, b = b, a + b
else:
    print("Entrada no válida. introduce un número entero positivo.")
```

```python
import random

colores_definidos = []
print("Primero, introduce diez colores diferentes:")

# Bucle para obtener 10 colores del usuario
for i in range(1, 11):
    while True:
        color_input = input(f"Introduce el color número {i}: ").strip().lower()
        if color_input:
            if color_input not in colores_definidos:
                colores_definidos.append(color_input)
                break
            else:
                print("Ese color ya ha sido introducido! Por favor, introduce un color diferente.")
        else:
            print("El color no puede estar vacío. Inténtalo de nuevo.")

print(f"\n¡Excelente! Los colores definidos son: {', '.join(colores_definidos)}")
print("Ahora, piensa en uno de estos colores. Yo intentaré adivinarlo en un máximo de 10 intentos.")

intentos = 0
max_intentos = 10
adivinado_correctamente = False
colores_disponibles_para_adivinar = list(colores_definidos) # Copia la lista para modificarla

# Bucle principal del juego
while intentos < max_intentos:
    intentos += 1

    if not colores_disponibles_para_adivinar:
        print("Vaya, Ya no me quedan más colores para adivinar. Parece que me he rendido.")
        break

    suposicion = random.choice(colores_disponibles_para_adivinar)
    print(f"\nIntento {intentos}/{max_intentos}: ¿Es tu color '{suposicion}'?")

    respuesta_usuario = input("Escribe 'sí' si es correcto, o 'no' si no lo es: ").strip().lower()

    if respuesta_usuario == 'sí':
        print(f"¡Siiii! ¡Adiviné tu color '{suposicion}' en {intentos} intentos!")
        adivinado_correctamente = True
        break
    elif respuesta_usuario == 'no':
        print("Vaya, no es ese. Déjame intentarlo de nuevo.")
        # Elimina la suposición para no repetirla en futuros intentos
        colores_disponibles_para_adivinar.remove(suposicion)
    else:
        print("Respuesta no válida. Por favor, responde 'sí' o 'no'. Este intento cuenta igual.")

# Resultado final del juego
if not adivinado_correctamente:
    if intentos == max_intentos:
        print(f"\n¡Se acabaron los {max_intentos} intentos! No pude adivinar tu color.")
    # Si no se adivinó y el bucle terminó por otra razón (ej. colores_disponibles_para_adivinar vacío antes de 10 intentos)
    else:
        print("El juego terminó sin que se adivinara el color.")
```

```python
numero_str = input("Por favor, introduce un número entero para calcular la suma de sus dígitos: ")

if numero_str.isdigit() or (numero_str.startswith('-') and numero_str[1:].isdigit()):
    numero_abs = abs(int(numero_str))
    suma_digitos = 0
    if numero_abs == 0:
        suma_digitos = 0
    else:
        temp_numero = numero_abs
        while temp_numero > 0:
            digito = temp_numero % 10
            suma_digitos += digito
            temp_numero //= 10
    print(f"La suma de los dígitos de {numero_str} es: {suma_digitos}")
else:
    print("Entrada no válida. Por favor, introduce un número entero.")
```

```python
import math

numero_str = input("Por favor, introduce un número entero para verificar si es primo: ")
if numero_str.isdigit() or (numero_str.startswith('-') and numero_str[1:].isdigit()):
    numero = int(numero_str)

    if numero <= 1:
        print(f"El número {numero} no es primo.")
    elif numero == 2:
        print(f"El número {numero} es primo.")
    elif numero % 2 == 0:
        print(f"El número {numero} no es primo (es par y mayor que 2).")
    else:
        es_primo = True
        for i in range(3, int(math.sqrt(numero)) + 1, 2):
            if numero % i == 0:
                es_primo = False
                break
        if es_primo:
            print(f"El número {numero} es primo.")
        else:
            print(f"El número {numero} no es primo.")
else:
    print("Entrada no válida. Por favor, introduce un número entero.")
```

```python
altura_str = input("Por favor, introduce la altura del triángulo (un número entero positivo): ")
if altura_str.isdigit():
    altura = int(altura_str)
    if altura > 0:
        print(f"\nTriángulo de asteriscos de altura {altura}:")
        for i in range(1, altura + 1):
            print('*' * i)
    else:
        print("Entrada no válida. Por favor, introduce un número entero positivo.")
else:
    print("Entrada no válida. Por favor, introduce un número entero positivo.")
```

```python
base_str = input("Introduce la base (un número entero): ")
exponente_str = input("Introduce el exponente (un número entero): ")
try:
    base = int(base_str)
    exponente = int(exponente_str)
    resultado = 1.0
    if exponente == 0:
        resultado = 1
    else:
        for _ in range(abs(exponente)):
            resultado *= base
        if exponente < 0:
            resultado = 1 / resultado
    print(f"El resultado de {base} elevado a {exponente} es: {resultado}")
except ValueError:
    print("Entrada no válida. Por favor, introduce números enteros para la base y el exponente.")
```
