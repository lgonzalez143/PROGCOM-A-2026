```python
numero = int(input("Introduce un número para ver su tabla de multiplicar: "))
multiplicador = 1
print(f"Tabla de multiplicar del {numero}:")
while multiplicador <= 10:
    resultado = numero * multiplicador
    print(f"{numero} x {multiplicador} = {resultado}")
    multiplicador += 1
```

```python
n = int(input("Introduce un número entero positivo (n): "))
suma = 0
contador = 1
while contador <= n:
    suma += contador
    contador += 1
print(f"La suma de los números naturales hasta {n} es: {suma}")
```

```python
num = int(input("Introduce un número entero no negativo para calcular su factorial: "))
if num < 0:
    print("El factorial no existe para números negativos.")
elif num == 0:
    print("El factorial de 0 es 1.")
else:
    factorial = 1
    contador = 1
    while contador <= num:
        factorial *= contador
        contador += 1
    print(f"El factorial de {num} es {factorial}.")
```

```python
n = int(input("Introduce la cantidad de números de Fibonacci que deseas ver: "))
a, b = 0, 1
contador = 0
if n <= 0:
    print("Introduce un número entero positivo.")
elif n == 1:
    print("Serie Fibonacci:")
    print(a)
else:
    print("Serie Fibonacci:")
    while contador < n:
        print(a)
        siguiente_fib = a + b
        a = b
        b = siguiente_fib
        contador += 1
```

```python
import random

# Paso 1: Definir los diez colores por el usuario
colores_definidos = []
print("Por favor, introduce diez colores diferentes (uno por uno):")
i = 0
while i < 10:
    color = input(f"Introduce el color {i + 1}: ").strip().lower()
    if color and color not in colores_definidos:
        colores_definidos.append(color)
        i += 1
    elif color in colores_definidos:
        print("Ese color ya ha sido introducido. Por favor, introduce uno diferente.")
    else:
        print("El color no puede estar vacío. Inténtalo de nuevo.")

# Verificar que se hayan introducido 10 colores
if len(colores_definidos) != 10:
    print("No se introdujeron 10 colores únicos. El juego no puede continuar.")
else:
    # Paso 2: Elige un color secreto al azar
    color_secreto = random.choice(colores_definidos)

    # Paso 3: Configura el juego
    intentos_maximos = 10
    intentos_restantes = intentos_maximos
    adivinado = False

    print("\n¡Bienvenido al juego de adivinar el color!")
    print(f"He elegido uno de los diez colores que definiste. Tienes {intentos_maximos} intentos para adivinarlo.")

    # Paso 4: Bucle while para los intentos de adivinanza
    while intentos_restantes > 0 and not adivinado:
        print(f"\nIntentos restantes: {intentos_restantes}")
        intento_usuario = input("Adivina el color: ").strip().lower()

        if intento_usuario == color_secreto:
            adivinado = True
            print(f"¡Felicidades! ¡Has adivinado el color secreto ({color_secreto.capitalize()})!")
        elif intento_usuario not in colores_definidos:
            print("Ese color no está en la lista de colores que definiste. ¡Intenta de nuevo con uno de ellos!")
        else:
            print("¡Incorrecto! Ese no es el color secreto.")

        intentos_restantes -= 1

    # Paso 5: Mensaje final después del bucle
    if not adivinado:
        print(f"\n¡Se te acabaron los intentos! El color secreto era: {color_secreto.capitalize()}.")
```

```python
num = int(input("Introduce un número entero para calcular la suma de sus dígitos: "))
suma_digitos = 0
numero_original = num
if num < 0:
    num = abs(num) #
if num == 0:
    suma_digitos = 0
else:
    while num > 0:
        digito = num % 10
        suma_digitos += digito
        num //= 10
print(f"La suma de los dígitos de {numero_original} es: {suma_digitos}")
```

```python
num = int(input("Introduce un número entero para verificar si es primo: "))
if num <= 1:
    print(f"{num} no es un número primo.")
elif num == 2:
    print(f"{num} es un número primo.")
else:
    es_primo = True
    divisor = 2
    while divisor < num:
        if num % divisor == 0:
            es_primo = False
            break
        divisor += 1
    if es_primo:
        print(f"{num} es un número primo.")
    else:
        print(f"{num} no es un número primo.")
```

```python
base = int(input("Introduce la base (un número entero): "))
exponente = int(input("Introduce el exponente (un número entero no negativo): "))
resultado = 1
if exponente < 0:
    print("Lo siento, este programa solo calcula potencias con exponentes no negativos.")
elif exponente == 0:
    print(f"{base} elevado a {exponente} es: 1")
else:
    contador = 0
    while contador < exponente:
        resultado *= base
        contador += 1
    print(f"{base} elevado a {exponente} es: {resultado}")
```
