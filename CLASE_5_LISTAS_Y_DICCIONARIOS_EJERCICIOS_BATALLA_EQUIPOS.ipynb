````python
##Ejercicio 2
# votos de los 5 candidatos
votos = [0, 0, 0, 0, 0]

n = int(input("Cuantos usuarios votan: "))

# primera votación
for i in range(n):
 voto = int(input("Vote por un candidato (1-5): "))

if voto >= 1 and voto <= 5:
 votos[voto-1] += 1
else:
 print("Voto invalido")

# mostrar votos
print("Resultados primera vuelta:")
for i in range(5):
 print("Candidato", i+1, ":", votos[i], "votos")

# encontrar los dos mayores
primero = votos.index(max(votos))
votos_temp = votos.copy()
votos_temp[primero] = -1
segundo = votos_temp.index(max(votos_temp))

print("Pasan a segunda vuelta:")
print("Candidato", primero+1)
print("Candidato", segundo+1)

# segunda votación
votos2 = [0, 0]

print("Segunda vuelta")

for i in range(n):
 voto = int(input(f"Vote por {primero+1} o {segundo+1}: "))

if voto == primero+1:
 votos2[0] += 1
elif voto == segundo+1:
 votos2[1] += 1
else:
 print("Voto invalido")

# resultados finales
print("Resultados finales:")

print("Candidato", primero+1, ":", votos2[0], "votos")
print("Candidato", segundo+1, ":", votos2[1], "votos")

if votos2[0] > votos2[1]:
 print("Ganador: Candidato", primero+1)
else:
 print("Ganador: Candidato", segundo+1)
````

#FUNCIÓN
Es un pedazo de código que se ejecuta aparte. A veces se conoce como copiado dinámico.

Ventajas de implementar las funciones:
1. Vuelve el dódigo mas limpio
2. Elimina el código espaguetti
3. Establece un orden de búsqueda.
4. Permite la reutilización del código

Sintaxis
```
 def nombre_de_la_función(parámetros):
  Aqui va todo lo que desees ejecutar
  return algo
```
Existen 4 tipos de funciones
1. Sin parámetros, sin retorno
2. Sin parámetros, con retorno
3. Con parámetros, sin retorno
4. Con parámetros, con retorno

````python
def menu():
    Comida = ["combo 1","combo 2", "combo 3", "combo 4","papas","jugos naturales","cerveza","agua","gaseosas"]
    precios = [20000,25000,18000,24500,3000,7000,3500,4000,2500]
    for i in range(len(Comida)):
        print(Comida[i], "-", precios[i])
    return Comida, precios


total_dia = 0
factura = []

user = "s"

while user == "s":
    Comida, precios = menu()
    r = "s"

    while r == "s":
        x = input("¿Qué desea?: ")

        if x in Comida:
            cantidad = int(input(f"¿Cuántas unidades de {x} quieres pedir?: "))
            precio_unitario = precios[Comida.index(x)]
            subtotal = precio_unitario * cantidad

            total_dia += subtotal


            factura.append((x, cantidad, precio_unitario, subtotal))

        else:
            print("Producto no válido")

        r = input("¿Desea algo más? (s/n): ")

    user = input("¿Hay más usuarios? (s/n): ")


print("\n FACTURA ")
for item in factura:
    nombre, cantidad, precio, subtotal = item
    print(f"{nombre} x{cantidad} - ${precio} = ${subtotal}")

print("///////////")
print(f"TOTAL DEL DÍA: ${total_dia}")
````
